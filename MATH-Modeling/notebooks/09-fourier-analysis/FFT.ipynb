```python
import numpy as np

def dft_matrix(n):
    k = np.arange(n)
    m = k.reshape((n, 1))
    A = np.exp(-2j * np.pi * k * m / n)
    return A

print(dft_matrix(2))
print(dft_matrix(4))
print(dft_matrix(8))
```

```python

# Define the block matrix D1
D1 = np.array([[1, 0, 0, 0],
                [0, (1-1j)/np.sqrt(2), 0, 0],
                [0, 0, -1j, 0],
                [0, 0, 0, (-1-1j)/np.sqrt(2)]])

# identity matrix I
I = np.identity(4)

# A8 
A8 = np.block([[I, D1], [-D1, np.zeros((4, 4))]])

# decompose to A4
A4 = np.block([[I, D1], [-D1, np.zeros((4, 4))]])[:4, :4]


print("A4:\n", A4)
print("\nA8:\n", A8)
```

# Exercise 4

**Base Case:**
For a 2 × 2 matrix, the decomposition is straightforward.

**Inductive Step:**

Assume that the decomposition process works for matrices of size $\frac{n}{2}$ x $\frac{n}{2}$, where n is a power of 2.

Divide the n x n matrix by 2 into four quadrants. After decomposing each quadrant, we obtain four smaller matrices of size $\frac{n}{2}$ x $\frac{n}{2}$. Then, we can combine these four smaller matrices into a single matrix of size $\frac{n}{2}$ x $\frac{n}{2}$. Thus, the $\frac{n}{2}$ x $\frac{n}{2}$ matrix is the decomposition of the original n x n matrix after one iteration. Then, we can repeat the reduction on the $\frac{n}{2}$ x $\frac{n}{2}$ matrix until we reach matrices of size 2x2, which represents the base case.

The number of iterations required is $log_2(n)$, as the matrix size is halved in each iteration.

```python
import matplotlib.pyplot as plt
x = np.linspace(0, 2*np.pi, 128)

def sine_wave(x):
    return np.sin(x)

def cosine_wave(x):
    return np.cos(x)

def square_wave(x):
    return np.where(x < np.pi, 1, -1)

def sawtooth_wave(x):
    return np.where(x < np.pi, x/np.pi, x/np.pi - 2)

random_array = np.random.default_rng(2024).uniform(size=128) - 0.5

fft_sine = np.fft.fft(sine_wave(x))
fft_cosine = np.fft.fft(cosine_wave(x))
fft_square = np.fft.fft(square_wave(x))
fft_sawtooth = np.fft.fft(sawtooth_wave(x))
fft_random = np.fft.fft(random_array)

freq_sine = np.abs(fft_sine)
freq_cosine = np.abs(fft_cosine)
freq_square = np.abs(fft_square)
freq_sawtooth = np.abs(fft_sawtooth)
freq_random = np.abs(fft_random)

plt.figure()
plt.plot(freq_sine)
plt.title('Sine Wave')

plt.figure()
plt.plot(freq_cosine)
plt.title('Cosine Wave')

plt.figure()
plt.plot(freq_square)
plt.title('Square Wave')

plt.figure()
plt.plot(freq_sawtooth)
plt.title('Sawtooth Wave')

plt.figure()
plt.plot(freq_random)
plt.title('Random Array')

plt.show()
```
