```python
import numpy as np
import matplotlib.pyplot as plt

# Define the DFT function
def compute_dft(f):
    N = len(f)
    n = np.arange(N)
    k = n.reshape((N, 1))
    e = np.exp(-2j * np.pi * n * k / N)
    return np.dot(e, f)

# Define the functions and compute their DFTs
N = 128
x = np.linspace(0, 2 * np.pi, N)

# Evaluate the functions
functions = {
    'sin': np.sin(x),
    'cos': np.cos(x),
    'square': np.where(x < np.pi, 1, -1),
    'sawtooth': np.where(x < np.pi, x/np.pi, x/np.pi - 2),
    'random': np.random.default_rng(2024).uniform(size=N) - 0.5
}

# Plot the magnitude of the DFT coefficients
plt.figure(figsize=(15, 10))
for i, (name, func) in enumerate(functions.items(), 1):
    F = compute_dft(func)
    plt.subplot(3, 2, i)
    plt.plot(np.abs(F))
    plt.title(f"DFT of {name}")
plt.show()
```

```python

def dft(f):
    N = len(f)
    F = np.fft.fft(f)
    return F

def plot_spectrum(F):
    k = np.arange(N)
    plt.plot(k, np.abs(F), '-')
    plt.show()

# Function definitions    
def sin_wave(x):
    return np.sin(x)

def cos_wave(x):
    return np.cos(x)

def square_wave(x):
    return 1 * (x < np.pi) - 1 * (x >= np.pi)

def sawtooth_wave(x):
    return x/np.pi * (x < np.pi) + (x/np.pi - 2) * (x >= np.pi)

def random_array(size):
    return np.random.default_rng(2024).uniform(size=size) - 0.5

# Evaluate functions on grid and compute DFT
N = 128
x = np.linspace(0, 2*np.pi, N)

for f in [sin_wave, cos_wave, square_wave, sawtooth_wave, random_array]:
    y = f(x) if f != random_array else f(N)
    Y = dft(y)
    plot_spectrum(Y)
```
