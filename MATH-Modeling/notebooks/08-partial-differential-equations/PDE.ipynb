$u_t = \frac{\partial}{\partial t} (e^{-0.2t} sin(\pi x)) = -0.2e^{-0.2t} sin(\pi x)$

$u_{xx} = \frac{\partial^2}{\partial x^2} (e^{-0.2t} sin(\pi x)) = -\pi^2 e^{-0.2t} sin(\pi x)$

Then, substitute $u_t = Dku_{xx}$

$-0.2e^{-0.2t} sin(\pi x) = Dk(-\pi^2 e^{-0.2t} sin(\pi x))$

Thus, $-0.2 = Dk(-\pi^2)$

$D = \frac{0.2}{k\pi^2}$, which represents $D$ in the term of $k$.

```python
import numpy as np
import matplotlib.pyplot as plt

u = lambda t, x: np.exp(-0.2 * t) * np.sin(np.pi * x)
x = np.linspace(0, 1, 100)
t = np.linspace(0, 10, 16)

fig, axes = plt.subplots(4, 4, figsize=(10, 8), sharey=True)
for ax, time in zip(axes.flat, t):
    ax.plot(x, u(time, x), "b")
    ax.grid(alpha=0.3)
    ax.set_ylim(0, 1)
    ax.set_title(f"t = {time:.2f}", fontsize=9)
for ax in axes[-1, :]:
    ax.set_xlabel("x")
for ax in axes[:, 0]:
    ax.set_ylabel("u")
fig.tight_layout()
plt.show()
```

```python
from ipywidgets import interactive

u = lambda t, x: np.exp(-0.2*t) * (1*np.sin(1*np.pi*x))
x = np.linspace(0,1,100)

def plotter(T):
  plt.plot(x , u(T,x), 'b')
  plt.grid()
  plt.ylim(0,1)
  plt.show()

interactive_plot = interactive(plotter, T=(0,20,0.1))
interactive_plot
```

```python

fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(projection="3d")
u = lambda t, x: np.exp(-0.2 * t) * np.sin(np.pi * x)
x = np.linspace(0, 1, 25)
t = np.linspace(0, 10, 25)
T, X = np.meshgrid(t, x)
ax.plot_wireframe(T, X, u(T, X))
ax.set_xlabel("time")
ax.set_ylabel("x")
ax.set_zlabel("u(t, x)")
plt.show()
```

```python
from matplotlib import animation

u = lambda t, x: np.exp(-0.2 * t) * np.sin(np.pi * x)
x = np.linspace(0, 1, 25)
t = np.linspace(0, 10, 101)

fig, ax = plt.subplots()
plt.close(fig)
ax.grid(alpha=0.3)
ax.set_xlabel("x")
ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
(frame,) = ax.plot([], [], linewidth=2, linestyle="--")

def animator(n):
    ax.set_title(f"t = {t[n]:.2f}")
    frame.set_data(x, u(t[n], x))
    return (frame,)

anim = animation.FuncAnimation(fig, animator, frames=range(len(t)), interval=100)
plt.rcParams["animation.html"] = "jshtml"
anim
```
