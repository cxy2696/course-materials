Exercise A.12

```python

def g(n):
    return (n**2 + n + 41)

outputs = [g(n) for n in range(0, 40)]
outputs
```

```python
for n in range(0, 42):
    for i in range(2, g(n)):
        if g(n) % i == 0:
            print(f"g({n}) = {g(n)} is not a prime number.")
```

```python
# a 
a = np.arange(1, 11, 1)
print(a)

square_a = a **2
print(square_a)
```

```python
# b
b = np.identity(10)
print(b)

b[0, 9] = 5
print("\n", b)
```

```python
# c
product_sqr = np.dot(b, square_a.T)
print(product_sqr)
```

```python
# d
b[-1, :] = 3
b[:, 2] = 7
print(b)

matrix = np.linalg.matrix_power(b, 5)
print("\n", matrix)
```

Exercise A.20

```python
def f(x):
    return x**2

def g(x):
    return x**3
    
def h(x):
    return x**4

x = np.linspace(0, 1, 400) # domain

# subplots
fig, axs = plt.subplots(1, 3, figsize = (30, 10))

# f(x) plot
axs[0].plot(x, f(x)) 
axs[0].set_title("f(x) = x^2")
axs[0].set_xlabel("x")
axs[0].set_ylabel("y")
axs[0].grid(True)

# g(x)
axs[1].plot(x, g(x))
axs[1].set_title("g(x) = x^3")
axs[1].set_xlabel("x")
axs[1].set_ylabel("y") 
axs[1].grid(True)

# h(x)
axs[2].plot(x, h(x))
axs[2].set_title("h(x) = x^4")
axs[2].set_xlabel("x")
axs[2].set_ylabel("y")
axs[2].grid(True)

fig.suptitle("Plots of Given Functions")
```
