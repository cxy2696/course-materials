```python
import numpy as np

L = np.array([[1, 0, 0],
              [4, 1, 0],
              [7, 2, 1]], dtype=float)
b = np.array([1, 0, 2], dtype=float)

y = np.zeros_like(b)
for i in range(b.size):
    y[i] = b[i] - L[i, :i] @ y[:i]
print(y)
```

<h1>4.29</h1>

```python
def lsolve(L, b):
    """Forward substitution for a unit lower-triangular system."""
    L = np.asarray(L, dtype=float)
    b = np.asarray(b, dtype=float).ravel()
    y = np.zeros(b.size)
    for i in range(b.size):
        y[i] = b[i] - L[i, :i] @ y[:i]
    return y


print(lsolve(L, b))
```

<h1>4.32</h1>

```python
U = np.array([[1, 2, 3],
              [0, -3, -6],
              [0, 0, -9]], dtype=float)
y = np.array([1, -4, 3], dtype=float)

x = np.zeros_like(y)
for i in range(y.size - 1, -1, -1):
    x[i] = (y[i] - U[i, i + 1:] @ x[i + 1:]) / U[i, i]
print(x)
```

<h1>4.33</h1>

```python
def usolve(U, y):
    """Back substitution for an upper-triangular system."""
    U = np.asarray(U, dtype=float)
    y = np.asarray(y, dtype=float).ravel()
    x = np.zeros(y.size)
    for i in range(y.size - 1, -1, -1):
        x[i] = (y[i] - U[i, i + 1:] @ x[i + 1:]) / U[i, i]
    return x


print(usolve(U, y))
```

```python
# Check the two triangular solvers on a random 5x5 system.
rng = np.random.default_rng(2024)
n = 5
L = np.tril(rng.random((n, n)), -1) + np.eye(n)
U = np.triu(rng.random((n, n)))
x_exact = rng.random(n)
b = L @ (U @ x_exact)

x = usolve(U, lsolve(L, b))
print(f"residual: {np.linalg.norm(x - x_exact):.2e}")
```
