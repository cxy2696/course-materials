np.linalg.svd() in Python: https://numpy.org/doc/stable/reference/generated/numpy.linalg.svd.html


```python
import numpy as np

def mySVD(A): # plug in a matrix: try to allow a rectangular matrix
    
    # compute A^T A and find its eigenvalues and eigenvectors
    eigenvalues, eigenvectors = np.linalg.eig(A.T @ A)

    # sort the eigenvalues in descending order, and sort the eigenvectors accordingly
    sorted_indices = np.argsort(eigenvalues)[::-1] # from the end towards the first taking each element
    V = eigenvectors[:, sorted_indices]
    
    sorted_eigenvalues = eigenvalues[sorted_indices]

    # compute Sigma, which has the square roots of the eigenvalues on the diagonal
    Sigma = np.sqrt(np.abs(sorted_eigenvalues))
    
    Sigma_output = np.zeros_like(A)
    np.fill_diagonal(Sigma_output, Sigma)
    
    # compute U 
    U = A @ V / Sigma
    

    return U, Sigma_output, V

m = 3
n = 2
A = np.random.randn(m, n)

# apply SVD function
U, Sigma, V = mySVD(A)

# use NumPy's svd function
U_np, Sigma_np, V_np = np.linalg.svd(A)

# convert the Sigma output array to a mxn matrix
Sigma_matrix = np.zeros_like(A)
np.fill_diagonal(Sigma_matrix, Sigma_np)

V_np = V_np.T # NumPy returns V transpose

print("mySVD function results:")
print("U:\n", U)
print("\nSigma:\n", Sigma)
print("\nV:\n", V)
print('----------------------------------------------')
print("NumPy's SVD Results:")
print("U:\n", U_np)
print("\nSigma:\n", Sigma_matrix)
print("\nV:\n", V_np)
```
