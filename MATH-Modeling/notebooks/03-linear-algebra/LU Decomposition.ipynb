<h1>4.22</h1>

```python
L1 = np.array([[1, 0, 0], [-4, 1, 0], [0, 0, 1]], dtype=float)
L2 = np.array([[1, 0, 0], [0, 1, 0], [-7, 0, 1]], dtype=float)
L3 = np.array([[1, 0, 0], [0, 1, 0], [0, -2, 1]], dtype=float)

for name, L in (("L1", L1), ("L2", L2), ("L3", L3)):
    print(f"{name}^-1 =\n{np.linalg.inv(L)}\n")
```

<h1>4.23</h1>

```python
L1_inv = np.linalg.inv(L1)
L2_inv = np.linalg.inv(L2)
L3_inv = np.linalg.inv(L3)

L_inv = L1_inv @ L2_inv @ L3_inv
print(L_inv)
```

<h1>The LU Factorization</h1>

```python
def myLU(A):
    """Doolittle LU factorization without pivoting."""
    A = np.asarray(A, dtype=float)
    n = A.shape[0]
    L = np.identity(n)
    U = A.copy()
    for j in range(n - 1):
        if U[j, j] == 0:
            raise ValueError("zero pivot: partial pivoting would be required")
        multipliers = U[j + 1:, j] / U[j, j]
        U[j + 1:, j:] -= multipliers[:, None] * U[j, j:]
        L[j + 1:, j] = multipliers
    return L, U
```

<h1>4.24</h1>

```python
A_2 = np.array([[2, 6, 9],
                [-6, 8, 1],
                [2, 2, 10]], dtype=float)

L, U = myLU(A_2)
print("L =\n", L)
print("U =\n", U)
print("L @ U =\n", L @ U)
print("max |A - LU| =", np.max(np.abs(A_2 - L @ U)))
```

<h1>4.25</h1>

```python
rng = np.random.default_rng(2024)
A = rng.standard_normal((10, 10))
L, U = myLU(A)
print("max |A - LU| for a random 10x10 matrix:",
      np.max(np.abs(A - L @ U)))
```
