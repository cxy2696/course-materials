```python
import numpy as np
import matplotlib.pyplot as plt

f = lambda x: np.sin(x) * (1-x) # lambda function of sin(x)(1-x)

exact = -np.sin(1) #  Compute the exact value of the derivative of f(x) at x=1 using the chain rule.

# Generate an array of h values l approximating the derivative.
# decreasing powers of 2, starting from 2^(-1) to 2^(-10)
H = 2.0**(-np.arange(1,10)) 


AbsPctError = [] # # Initialize a list to store the absolute percentage errors for each h.


for h in H:
    approx = (f(1+h) - f(1)) / h  # Compute the approximate derivative
    
    # Add the absolute percentage error to the list.
    AbsPctError.append( np.abs( (approx - exact)/exact ) )
    
    # If this is the first iteration, print the error; otherwise, calculate the error reduction.
    if h==H[0]:
      print("h=",h,"\t Absolute Pct Error=", AbsPctError[-1])
      
    else:
      err_reduction_factor = AbsPctError[-2]/AbsPctError[-1]
      print("h=",h,"\t Absolute Pct Error=", AbsPctError[-1],
              "with error reduction",err_reduction_factor)
      
plt.loglog(H,AbsPctError,'b-*') # log-log plot helps to assess the order of the convergence rate, the value of a slope
plt.grid()
plt.show()
```

<h1>3.17</h1>

a. we need a value of the function at the point that is a small step h ahead. At the last point in the interval $[0, 2\pi]$, there exists no subsequent point.

```python
# b.


def FirstDeriv(f,a,b,N):
    x = np.linspace(a,b,N+1)  # Generate N+1 points between a and b. N subintervals means N+1 points
    
    h = x[1] - x[0]           # Calculate the step size h based on the spacing of the linspace array
    
    df = []                   # Initialize an empty list to store the derivative approximations
    
    for j in np.arange(len(x)-1):  # Iterate over the array x until the second last element
        approx = (f(x[j+1]) - f(x[j])) / h  # Approximate the derivative using forward difference
        
        df.append(approx)  # Add the calculated approximation to the list df
    
    return df  # Return the list of approximated derivatives
```

```python
# c.  

f = lambda x: np.sin(x)  # Function for which we are finding the derivative
exact_df = lambda x: np.cos(x)  # The exact derivative of f

a = 0  # Start of the interval
b = 2*np.pi  # End of the interval
N = 100  # Number of subintervals

x = np.linspace(a,b,N+1)  # Generate N+1 points for N subintervals
df = FirstDeriv(f,a,b,N)  # Get the numerical approximation of the first derivative


# In the next line we plot three curves: 
# 1) the function f(x) = sin(x)
# 2) the function f'(x) = cos(x)
# 3) the approximation of f'(x)

# Plot the function, its exact derivative, and the numerical approximation
plt.plot(x,f(x),'b-',x[:-1],exact_df(x[:-1]),'r-',x[:-1],df,'k.')
plt.grid()
plt.legend(['f(x) = sin(x)', 'exact first deriv', 'approx first deriv'])
plt.show()
```
