```python
import numpy as np
import matplotlib.pyplot as plt


def secondDeriv(f, a, b, N):
    x = np.linspace(a,b, N+1)
    h = x[1] - x[0]
    
    d2f = (f(x[2:]) - 2*f(x[1:-1]) + f(x[:-2]))/h**2
    
    return x, d2f 

f = lambda x: np.sin(x)*(1-x)
fPrimeExact = lambda x: np.cos(x)*(1-x) - np.sin(x)
d2f = lambda x: -np.sin(x)*(1-x) - np.cos(x) - np.cos(x)

x_approx, d2f_approx = secondDeriv(f, 0, 5, 100)

plt.plot(x_approx,f(x_approx), color = "blue")
plt.plot(x_approx,d2f(x_approx), color = "red")
plt.plot(x_approx[1:-1],d2f_approx, linestyle = "dashdot", color = "green")
```

```python
def second_order(f, a, b, N):
    x = np.linspace(a, b, N+1)
    h = x[1] - x[0]
    
    approx = (f(x[2:]) - 2*f(x[1:-1]) + f(x[:-2])) / h**2
    
    return x[1:-1], approx


f = lambda x: np.sin(x) - x * np.sin(x)
fprime = lambda x: np.cos(x) - np.sin(x) - x * np.cos(x)
df = lambda x: -np.sin(x) - np.cos(x) + np.sin(x) - x * np.sin(x)


x, second_approx = second_order(f, 0, 10, 100)
second_exact = df(x)
errors = np.abs(second_approx - second_exact)
max_error = np.max(errors)


N_values = np.arange(2, 1000)
h_values = []
max_errors = []

for N in N_values:
    h_values.append(10/N)
    x_values, second_derivative = second_order(f, 2, 20, N)
    second_derivative_exact = df(x_values) 
    errors = np.abs(second_derivative - second_derivative_exact)
    max_errors.append(np.max(errors))


plt.loglog(h_values, max_errors)
plt.xlabel('Step size h')
plt.ylabel('Maximum Error')
plt.show()
```
