<h1>1.26</h1>

Euler's number $e = \sum_{n=0}^\infty \frac{1}{n!} = 2.718$ approximately. <br>
The first derivative of $f(x)$ is $f'(x) = e^x$<br>
$f'(0) = 1$<br>
The line equation $g(x)$ passing through the point $(0,1)$ with a slope of 1 is $g(x) = f(0) + f'(0)(x-0) = 1 + x$

```python
x = np.linspace(-1.5, 1.5, 400)

def g_1(X):
    return np.ones_like(x)

plt.plot(x, g_1(x))
```

```python

x = np.linspace(-1.5, 1.5, 400)

def f(x):
    return np.exp(x)

def g_2(x):
    return 1 + x

plt.plot(x, f(x), label = "f(x) = e^x", color = "r", linestyle = "dashed")
plt.axhline(y = 1, label = "constant", color = "b", linestyle = "dashdot")
plt.plot(x, g_2(x), label = "linear", color = "black")

plt.title("Two approximations of e^x at x = 0")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.legend()
```

<h1>1.27</h1>

The quadratic approximation is $ y = 1 + x + \frac{1}{2} x^2$<br>
The function $f(x)$'s value, first derivative, and second derivative at $x = 0$ are the same.

```python
def y(x):
    return 1 + x + 0.5*x**2

plt.plot(x, y(x), label = "quadratic")

plt.plot(x, f(x), label = "f(x) = e^x", color = "r", linestyle = "dashed")
plt.axhline(y = 1, label = "constant", color = "b", linestyle = "dashdot")
plt.plot(x, g_2(x), label = "linear", color = "black")
plt.title("Approximations of e^x at x = 0")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.legend()
```

The cubic approximation is $y = 1 + x + \frac{1}{2} x^2 + \frac{1}{6} x^3$ <br>
We know this functions matches the first, second, and third derivative of $f(x)$ at $x=0$ since all values are the same.

```python
def y_2(X):
    return 1 + x + 0.5*x**2 + (1/6)*x**3

plt.plot(x, y_2(x), label = "cubic")

plt.plot(x, y(x), label = "quadratic")

plt.plot(x, f(x), label = "f(x) = e^x", color = "r", linestyle = "dashed")
plt.axhline(y = 1, label = "constant", color = "b", linestyle = "dashdot")
plt.plot(x, g_2(x), label = "linear", color = "black")
plt.title("Approximations of e^x at x = 0")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.legend()
```
