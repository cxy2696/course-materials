```python
xdata = np.array([0, 1, 2, 3]) # import x values
ydata = np.array([1.07, 3.9, 14.8, 26.8]) # y values

def SSRes(parameters): # function for fitting curve
  # In the next line of code we want to build our 
  # quadratic approximation y = ax^2 + bx + c
  # We are sending in a list of parameters so 
  # a = parameters[0], b = parameters[1], and c = parameters[2]

  # Define the quadratic equation as approximations to y values
  yapprox = parameters[0]*xdata**2 + \
            parameters[1]*xdata + \
            parameters[2]
            
  residuals = np.abs(ydata-yapprox) # computes the absolute differences between the actual y values and approximations
  return np.sum(residuals**2) # return the sum of residual squares, which we need to minimize


BestParameters = minimize(SSRes,[2,2,0.75]) # call SSRes function and minimize the return with an initial guess of [2,2,0.75] for a, b, c
print("The best values of a, b, and c are: \n",BestParameters.x)
# If you want to print the diagnositc then use the line below:
# print("The minimization diagnostics are: \n",BestParameters)
plt.plot(xdata,ydata,'bo',markersize=5)
x = np.linspace(0,4,100)
y = BestParameters.x[0]*x**2 + \
    BestParameters.x[1]*x + \
    BestParameters.x[2]
plt.plot(x,y,'r--')
plt.grid()
plt.xlabel('x')
plt.ylabel('y')
plt.title('Best Fit Quadratic')
plt.show()
```
