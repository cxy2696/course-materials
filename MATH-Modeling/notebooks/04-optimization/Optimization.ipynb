```python
import numpy as np

def derivative_free_optimization(f, x0, dx, tol):
    x = x0
    
    while np.abs(f(x)) > tol:
        f_left = f(x - dx)
        f_right = f(x + dx)
        
        if f_left > f(x):
            x = x - dx
            
        elif f_right > f(x):
            x = x + dx
            
        else: # find the maximum
            break
        
        dx = -dx/2    
        
    return x, f(x) 


f = lambda x: x*(20-2*x)**2

x_max, f_max = derivative_free_optimization(f, 3, 0.01, 1e-6)

print(f"Max x value is {x_max} with volume of {f_max}")
```

<h1>3.53</h1>

```python
def gradient_descent(f, df, x0, alpha, tol):
    x = x0
    
    while np.abs(df(x)) > tol:
        f_prime = df(x)
        
        x_new = x + alpha*f_prime
        
        x = x_new 
        
    return x, f(x)

f = lambda x: x*(20-2*x)**2
df = lambda x: (20-2*x)**2 + 8*x**2 - 80*x

x_max, f_max = gradient_descent(f, df, 1, 0.01, 1e-6)
print(x_max, f_max)
```

```python
import random

def monte_carlo(f, x_min, x_max, y_min, y_max, num_samples):
    best_solution = None
    best_value = float("inf")  # or Infinity if minimizing
    
    for _ in range(num_samples):
        x = random.uniform(x_min, x_max)
        y = random.uniform(y_min, y_max)
        value = f(x, y)
        
        if value < best_value:
            best_solution = (x, y)
            best_value = value
    return best_solution, best_value


f = lambda x, y: np.sin(x)*np.cos(y) + 0.1*x**2

result = monte_carlo(f, 0, 10, 0, 10, 100)

print(result)

```
