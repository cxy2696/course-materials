```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(2024)   # fixed seed: the saved output is reproducible
nums = 1_000_000

x = rng.uniform(-5, 5, size=nums)
max_pdf = 1 / np.sqrt(2 * np.pi)    # height of the standard normal density
y = rng.uniform(0, max_pdf, size=nums)
```

```python
# Keep the samples that fall under the normal density curve.
pdf = np.exp(-x**2 / 2) / np.sqrt(2 * np.pi)
accepted = x[y < pdf]
print(f"accepted {accepted.size:,} of {nums:,} samples "
      f"({accepted.size / nums:.1%})")
```

```python
plt.hist(accepted, bins=100, density=True, alpha=0.7)
plt.title("Samples from the standard normal distribution (accept-reject)")
plt.show()
```

```python
print(f"sample minimum: {accepted.min():.3f}")
print(f"sample maximum: {accepted.max():.3f}")
```
