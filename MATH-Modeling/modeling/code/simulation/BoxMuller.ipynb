```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(2024)
nums = 100_000
u1 = rng.uniform(size=nums)
u2 = rng.uniform(size=nums)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 3.5))
axes[0].hist(u1, bins=100)
axes[0].set_title("u1 ~ Uniform(0, 1)")
axes[1].hist(u2, bins=100)
axes[1].set_title("u2 ~ Uniform(0, 1)")
plt.tight_layout()
plt.show()
```

```python
radius = np.sqrt(-2 * np.log(u1))
z1 = radius * np.cos(2 * np.pi * u2)
z2 = radius * np.sin(2 * np.pi * u2)

plt.hist(z1, bins=100, density=True, alpha=0.7)
plt.title("Box-Muller: first normal sample")
plt.show()
```

```python
plt.hist(z2, bins=100, density=True, alpha=0.7)
plt.title("Box-Muller: second normal sample")
plt.show()
```
