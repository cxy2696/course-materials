```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(2024)

rolls = rng.integers(1, 7, size=10)
print(rolls)
print(f"sum of the ten rolls: {rolls.sum()}")
```

```python
# 100 rows, 10 dice per row
intmatrix = rng.integers(1, 7, size=(100, 10))
intsums = intmatrix.sum(axis=1)
```

```python
plt.hist(intsums, bins=range(10, 71, 2), edgecolor="black")
plt.xlabel("sum of ten dice")
plt.ylabel("count")
plt.show()
```

```python
print(f"mean: {intsums.mean():.3f}")
print(f"standard deviation: {intsums.std():.3f}")
print(f"maximum: {intsums.max()}")
```
