```python
import numpy as np

rng = np.random.default_rng(2024)
nums = 100_000

def roll_dice(count):
    """Roll `count` fair six-sided dice."""
    return rng.integers(1, 7, size=(count, 2)).sum(axis=1)

first = roll_dice(nums)
wins = np.zeros(nums, dtype=int)
wins[np.isin(first, (7, 11))] = 1
wins[np.isin(first, (2, 3, 12))] = 0

# The remaining games continue until the point or a 7 is rolled.
point = first.copy()
active = ~np.isin(first, (2, 3, 7, 11, 12))

while active.any():
    idx = np.flatnonzero(active)
    total = roll_dice(idx.size)
    hit_point = total == point[idx]
    hit_seven = total == 7
    wins[idx[hit_point]] = 1
    wins[idx[hit_seven]] = 0
    active[idx[hit_point | hit_seven]] = False

print(f"empirical win probability: {wins.mean():.4f}")
```
