# Bet on Red or Black

```python
import numpy as np

rng = np.random.default_rng(2024)
nums = 1_000_000

outcomes = rng.integers(-1, 36, size=nums, endpoint=True)
winnings_red = np.where((outcomes > 0) & (outcomes % 2 == 0), 1, -1)

def longest_run(mask):
    """Length of the longest consecutive run of True values."""
    if not mask.any():
        return 0
    edges = np.flatnonzero(np.diff(np.r_[False, mask, False]))
    return (edges[1::2] - edges[::2]).max()

print(f"mean winnings (red/black bet): {winnings_red.mean():.4f}")
print(f"longest winning streak: {longest_run(winnings_red == 1)}")
print(f"longest losing streak:  {longest_run(winnings_red == -1)}")
```

# Bet on Green

```python
# Betting on a single green zero pays 17:1.
winnings_green = np.where(outcomes == 0, 17, -1)

print(f"mean winnings (green bet): {winnings_green.mean():.4f}")
print(f"longest winning streak: {longest_run(winnings_green == 17)}")
print(f"longest losing streak:  {longest_run(winnings_green == -1)}")
```
