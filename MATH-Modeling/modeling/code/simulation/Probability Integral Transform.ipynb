```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(2024)
nums = 10_000
u = rng.uniform(size=nums)

plt.hist(u, bins=50, alpha=0.7)
plt.title("Uniform(0, 1) samples")
plt.show()
```

```python
rate = 1.0
exponential = -np.log(u) / rate

plt.hist(exponential, bins=50, density=True, alpha=0.7)
plt.title("Exponential samples via the probability integral transform")
plt.show()
```
