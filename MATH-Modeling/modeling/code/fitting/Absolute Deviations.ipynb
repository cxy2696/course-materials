```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.optimize as optimize

speed = np.arange(17, dtype=float)
stopping = np.array([0, 3, 6, 11, 21, 32, 47, 65, 87, 112,
                     140, 171, 204, 241, 282, 325, 376], dtype=float)


def sum_absolute_deviations(a):
    """L1 objective for the model y = a x^2."""
    return np.sum(np.abs(stopping - a * speed**2))


result = optimize.minimize(sum_absolute_deviations, x0=1.0)
a = result.x[0]
print(f"fitted a = {a:.6f}, objective = {result.fun:.4f}")
```

```python
predict = a * speed**2
plt.scatter(speed, stopping, label="data")
plt.plot(speed, predict, color="orange", label="L1 fit")
plt.xlabel("speed")
plt.ylabel("stopping distance")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
residual = stopping - predict
plt.plot(speed, residual, "o-")
plt.axhline(0, color="black", lw=0.8)
plt.xlabel("speed")
plt.ylabel("residual")
plt.grid(alpha=0.3)
plt.show()
```
