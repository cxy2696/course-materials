```python
import matplotlib.pyplot as plt

xvals = range(1,11)  # This creates xvals as a list of numbers from 1 to 10.
yvals = [4, 11, 22, 35, 36, 80, 107, 140, 175, 215]  # Data

x_cubed = [x**3 for x in xvals]  # This technique is called "list comprehension."  It creates a list consisting of the cubes of the xvals list.

plt.scatter(x_cubed,yvals)
```

```python
plt.plot(x_cubed,yvals)
```

```python
plt.scatter(xvals,yvals)
```

```python
x_fourth = [x**4 for x in xvals]  

plt.scatter(x_fourth,yvals)
```

```python
x_sq = [x**2 for x in xvals]  

plt.plot(x_sq,yvals)
```
