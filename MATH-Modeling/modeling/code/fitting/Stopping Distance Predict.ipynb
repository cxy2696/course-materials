```python
import matplotlib.pyplot as plt

StoppingData = [0, 3, 6, 11, 21, 32, 47, 65, 87, 112, 140, 171, 204, 241, 282, 325, 376]
plt.plot(StoppingData)

k = 3.2

StopPredict = [0]*len(StoppingData)
StopPredict[0] = 0
for i in range(len(StoppingData)-1):
    StopPredict[i+1] = StopPredict[i] + k*i

plt.plot(StopPredict)
```
