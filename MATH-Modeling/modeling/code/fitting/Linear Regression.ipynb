```python
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.api as sm

speed = np.arange(17, dtype=float)
stopping = np.array([0, 3, 6, 11, 21, 32, 47, 65, 87, 112,
                     140, 171, 204, 241, 282, 325, 376], dtype=float)

# Regress stopping distance on speed^2 with an intercept.
X = sm.add_constant(speed**2)
model = sm.OLS(stopping, X).fit()
print(model.params)
print(f"R^2 = {model.rsquared:.4f}")
```

```python
fit = model.predict(X)

plt.scatter(speed, stopping, label="data")
plt.plot(speed, fit, color="orange", label="OLS fit")
plt.xlabel("speed")
plt.ylabel("stopping distance")
plt.legend()
plt.grid(alpha=0.3)
plt.show()

plt.plot(speed, stopping - fit, "o-")
plt.axhline(0, color="black", lw=0.8)
plt.xlabel("speed")
plt.ylabel("residual")
plt.grid(alpha=0.3)
plt.show()
```
