```python
import numpy as np
import matplotlib.pyplot as plt

# Define the range for x values
x_values = np.linspace(-6, 6, 100)

# Compute the boundaries for z1 = 0 and z2 = 0
# z1 = 0 boundary -> x1 = 2 * x2, rearranged to x2 = 0.5 * x1
z1_boundary_y = 0.5 * x_values

# z2 = 0 boundary -> x2 = 3 * x1
z2_boundary_y = 3 * x_values

# Create plot
plt.figure(figsize=(5, 5))
plt.plot(x_values, z1_boundary_y, 'r--', label=r'$z_1 = 0 \ (x_1 = 2x_2)$')  # z1 boundary as dotted line
plt.plot(x_values, z2_boundary_y, 'b--', label=r'$z_2 = 0 \ (x_2 = 3x_1)$')  # z2 boundary as dotted line

# Illustrative decision boundary line, placed at x1 = -1 for visualization
decision_boundary_x = np.full_like(x_values, -1)
plt.plot(decision_boundary_x, x_values, 'k-', label='Decision Boundary')

# Set plot limits and labels
plt.xlim(-6, 6)
plt.ylim(-6, 6)
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.grid(True, which='both')

# Add labels and legend
plt.xlabel(r'$x_1$')
plt.ylabel(r'$x_2$')
plt.title('Decision Boundaries for Neural Network')
plt.legend(loc='upper left')

# Display the plot
plt.show()
```

```python
import numpy as np

# Define matrix B with a "cross" of ones in the center
B = np.zeros((15, 15))
B[5:10, :] = 1  # Set the middle 5 rows to 1
B[:, 5:10] = 1  # Set the middle 5 columns to 1

# Define filter F2
F2 = np.array([
    [-2, -1, 0, 1, 2],
    [-2, -1, 0, 1, 2],
    [-2, -1, 0, 1, 2],
    [-2, -1, 0, 1, 2],
    [-2, -1, 0, 1, 2]
])

# Perform convolution without padding and with stride 1
output_size = 11  # Calculated as 15 - 5 + 1
output = np.zeros((output_size, output_size))

for i in range(output_size):
    for j in range(output_size):
        # Extract the 5x5 region of B
        region = B[i:i+5, j:j+5]
        # Compute the convolution as the sum of element-wise multiplication
        output[i, j] = np.sum(region * F2)

# Print the resulting output matrix
print("Output after convolving B with F2:")
print(output)
```
