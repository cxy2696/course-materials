```python
import numpy as np
import matplotlib.pyplot as plt

# Define price range
P = np.linspace(0, 10, 100)

# Supply and demand functions
Q_S = 18.5 + 0.7 * P  # Supply equation
Q_D = 34.5 - 1.8 * P  # Demand equation

# Equilibrium point from part (a)
P_eq = 6.4
Q_eq = 22.98

# Price ceiling at P = 7
P_ceiling = 7
Q_S_ceiling = 18.5 + 0.7 * P_ceiling  # Quantity supplied at price ceiling
Q_D_ceiling = 34.5 - 1.8 * P_ceiling  # Quantity demanded at price ceiling

# Plotting the supply and demand curves
plt.figure(figsize=(8, 6))
plt.plot(Q_S, P, label="Supply Curve", color="blue")
plt.plot(Q_D, P, label="Demand Curve", color="red")

# Mark equilibrium
plt.scatter(Q_eq, P_eq, color="green", label="Equilibrium ($6.4, 22.98)")

# Mark the price ceiling effect
plt.axhline(y=P_ceiling, color="purple", linestyle="--", label="Price Ceiling ($7)")
plt.scatter(Q_S_ceiling, P_ceiling, color="blue", marker="o", label="Supply at $7 (23.4)")
plt.scatter(Q_D_ceiling, P_ceiling, color="red", marker="o", label="Demand at $7 (21.9)")

# Labels and title
plt.xlabel("Quantity (Tcf)")
plt.ylabel("Price ($ per Tcf)")
plt.title("Effect of Price Ceiling on Natural Gas Market")
plt.legend()
plt.grid(True)

# Show the plot
plt.show()
```
