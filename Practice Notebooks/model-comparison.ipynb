```python
import pandas as pd
import numpy as np
data = pd.read_csv('data/preds.csv')
data.head()
```

```python
n = len(data)
h_A_acc = np.mean(data['truth'] != data['h_A'])
h_B_acc = np.mean(data['truth'] != data['h_B'])
alpha = 0.05
epsilon = np.sqrt(np.log(2/alpha)/(2*n))
h_A_con_int = (h_A_acc - epsilon, h_A_acc + epsilon)
h_B_con_int = (h_B_acc - epsilon, h_B_acc + epsilon)
print(h_A_con_int)
print(h_B_con_int)
```

```python
# Create the contingency table
# Wins and losses based on the comparison between h_A and h_B

# a: Both h_A and h_B are correct
a = ((data['truth'] == data['h_A']) & (data['truth'] == data['h_B'])).sum()

# b: h_A correct, h_B incorrect
b = ((data['truth'] == data['h_A']) & (data['truth'] != data['h_B'])).sum()

# c: h_A incorrect, h_B correct
c = ((data['truth'] != data['h_A']) & (data['truth'] == data['h_B'])).sum()

# d: Both h_A and h_B are incorrect
d = ((data['truth'] != data['h_A']) & (data['truth'] != data['h_B'])).sum()

# Calculate the McNemar test statistic
from scipy.stats import binom

# Set parameters for the binomial distribution
n = b + c  # total wins + losses
p = 0.5  # for null hypothesis

# Calculate cumulative probability P(W <= b)
p_wins = binom.cdf(b, n, p)

# Calculate cumulative probability P(W >= c), which is 1 - P(W < c)
p_losses = 1 - binom.cdf(c - 1, n, p)

print(a, b, c, d)
print(p_wins)
print(p_losses)
```
