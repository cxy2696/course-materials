```python
import matplotlib.pyplot as plt
import networkx as nx

# Create a data flow graph
data_flow_graph = nx.DiGraph()

# Define nodes and edges for the Data Flow Diagram
data_flow_nodes = [
    "Raw Dataset",
    "Data Preprocessing",
    "Augmented Dataset",
    "YOLO Model",
    "Multi-class Classification",
    "UI/UX Interface",
    "Classification Results"
]

data_flow_edges = [
    ("Raw Dataset", "Data Preprocessing"),
    ("Data Preprocessing", "Augmented Dataset"),
    ("Augmented Dataset", "YOLO Model"),
    ("YOLO Model", "Multi-class Classification"),
    ("Multi-class Classification", "UI/UX Interface"),
    ("UI/UX Interface", "Classification Results")
]

data_flow_graph.add_nodes_from(data_flow_nodes)
data_flow_graph.add_edges_from(data_flow_edges)

# Draw the Data Flow Diagram
plt.figure(figsize=(12, 6))
pos = nx.spring_layout(data_flow_graph, seed=42)
nx.draw(data_flow_graph, pos, with_labels=True, node_size=3000, node_color='lightblue', font_size=10, font_weight='bold', arrowsize=20)
plt.title("Data Flow Diagram", fontsize=14, fontweight='bold')
plt.show()

# Create an execution diagram graph
execution_graph = nx.DiGraph()

# Define nodes and edges for the Execution Diagram
execution_nodes = [
    "Setup Environment",
    "Load Dataset",
    "Preprocess Data",
    "Train YOLO Model",
    "Evaluate Model",
    "Run UI/UX",
    "Display Results"
]

execution_edges = [
    ("Setup Environment", "Load Dataset"),
    ("Load Dataset", "Preprocess Data"),
    ("Preprocess Data", "Train YOLO Model"),
    ("Train YOLO Model", "Evaluate Model"),
    ("Evaluate Model", "Run UI/UX"),
    ("Run UI/UX", "Display Results")
]

execution_graph.add_nodes_from(execution_nodes)
execution_graph.add_edges_from(execution_edges)

# Draw the Execution Diagram
plt.figure(figsize=(12, 6))
pos = nx.spring_layout(execution_graph, seed=42)
nx.draw(execution_graph, pos, with_labels=True, node_size=3000, node_color='lightgreen', font_size=10, font_weight='bold', arrowsize=20)
plt.title("Execution Diagram", fontsize=14, fontweight='bold')
plt.show()
```
